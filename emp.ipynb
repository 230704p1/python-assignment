```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df,x='age',y='salary',color='red',edgecolor='linen',alpha=0.5)
plt.title('Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```

 **Age vs Salary (Scatter Plot)**: This scatter plot shows no strong linear correlation between an employee's age and their salary. While there's a wide range of salaries across different ages, there doesn't appear to be a clear trend where salary consistently increases or decreases with age. Salaries seem to be distributed across various age groups without a distinct pattern.

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df,x='healthy_eating',y='active_lifestyle',color='red',edgecolor='linen',alpha=0.5)
plt.title('Healthy Eating vs Active Lifestyle')
plt.xlabel('Healthy Eating')
plt.ylabel('Active Lifestyle')
plt.show()
```

**Healthy Eating vs Active Lifestyle (Scatter Plot):** This plot displays the relationship between healthy eating scores and active lifestyle scores. Similar to the age vs salary plot, there isn't an obvious strong correlation. Points are scattered across the graph, suggesting that high healthy eating scores don't necessarily correspond to high active lifestyle scores, or vice-versa, within this dataset.

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='groups', palette='viridis')
plt.title('Count of Employees by Group')
plt.xlabel('Group')
plt.ylabel('Count')
plt.show()
```

**Count of Employees by Group (Count Plot/Bar plot):** This count plot clearly illustrates the distribution of employees across different groups (A, O, B, AB). Group 'A' appears to have the highest number of employees, followed by 'O' and 'B', with 'AB' having the fewest. This provides a quick overview of the group sizes within the dataset.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='salary', kde=True, color='skyblue')
plt.title('Distribution of Salary')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

**Distribution of Salary (Histogram):** The histogram of salary distribution, coupled with the Kernel Density Estimate (KDE) line, indicates that salaries in this dataset are not uniformly distributed. There appears to be a concentration of salaries in the lower to middle ranges, with fewer employees earning very high salaries. The distribution might be slightly skewed to the right, suggesting more employees earn less than the average, with a few outliers earning significantly more.
